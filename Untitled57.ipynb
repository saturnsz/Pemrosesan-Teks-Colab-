```python
import re

teks = "Vaksinasi && adalah pemberian Vaksin dalam rangka menimbulkan atau meningkatkan kekebalan seseorang !! secara aktif terhadap +=?? suatu penyakit, sehingga apabila suatu saat terpajan dengan penyakit [tersebut] tidak !! akan sakit atau hanya mengalami sakit ringan dan tidak menjadi sumber penularan"

# Menghapus tanda baca menggunakan Regular Expression (regex)
teks_bersih = re.sub(r'[^\w\s]', '', teks)

# Membersihkan spasi ganda akibat penghapusan karakter
teks_bersih = re.sub(r'\s+', ' ', teks_bersih).strip()

print(teks_bersih)
```

### Soal 1: Menghilangkan Tanda Baca pada Kalimat
Pada bagian ini, kita menggunakan *library* `re` (Regular Expression) bawaan dari Python untuk mendeteksi dan menghapus semua karakter tanda baca (seperti `&&`, `!!`, `+=??`, `[]`, dll).
- `re.sub(r'[^\w\s]', '', teks)` digunakan untuk membuang semua karakter selain huruf, angka, dan spasi.
- `re.sub(r'\s+', ' ', teks_bersih).strip()` digunakan untuk merapikan spasi ganda yang muncul setelah tanda baca dihapus agar kalimat kembali rapi.

```python
sent = ['Sebuah', 'bunga', 'diberikan', 'John', 'untuk', 'lady']
result = []

for word in sent:
    word_len = (word, len(word))
    result.append(word_len)

print(result)
```

### Soal 2: Menghitung Jumlah Karakter pada Tiap Kata
Pada sintaks ini, kita diberikan sebuah *list* berisi beberapa kata. Tujuan dari kode ini adalah:
1. Melakukan perulangan (`for word in sent`) untuk mengambil setiap kata di dalam *list*.
2. Menggunakan fungsi `len(word)` untuk menghitung jumlah huruf dari kata tersebut.
3. Menggabungkan kata dan jumlah hurufnya ke dalam bentuk *tuple* `(word, len(word))`.
4. Memasukkan *tuple* tersebut ke dalam *list* baru bernama `result` menggunakan metode `.append()`.
