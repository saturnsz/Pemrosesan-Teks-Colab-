```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.stem import PorterStemmer
```

```python
# Unduh resource NLTK jika belum tersedia
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)

# 1.a. Membuat variabel contoh_raw
contoh_raw = (
    "Python is an interpreted high-level general-purpose programming language. "
    "Its design philosophy emphasizes code readability with its use of significant indentation. "
    "Its language constructs as well as its object-oriented approach aim to help "
    "programmers write clear, logical code for small and large-scale projects"
)

print("--- VARIABEL CONTOH_RAW ---")
print(contoh_raw)
print("\n" + "="*50 + "\n")
```

```python
# 1.b. Menghitung matriks frekuensi kata dalam setiap kalimat dengan stopwords "English"
# Unduh resource punkt_tab yang hilang
nltk.download('punkt_tab', quiet=True)

def create_frequency_matrix(sentences):
    frequency_matrix = {}
    stop_words = set(stopwords.words("english"))
    ps = PorterStemmer()

    for sent in sentences:
        freq_table = {}
        words = word_tokenize(sent)
        for word in words:
            word = word.lower()
            word = ps.stem(word)
            if word in stop_words:
                continue

            if word in freq_table:
                freq_table[word] += 1
            else:
                freq_table[word] = 1

        frequency_matrix[sent[:15]] = freq_table

    return frequency_matrix

sentences = sent_tokenize(contoh_raw)
freq_matrix = create_frequency_matrix(sentences)

print("--- 1.b. MATRIKS FREKUENSI KATA (LATIHAN) ---")
for sent_key, freq in freq_matrix.items():
    print(f"Kalimat (slice): {sent_key}...")
    print(freq)
    print()

print("="*50 + "\n")
```

```python
# 2. Melakukan perhitungan skor rata-rata (TF) dari contoh_raw diatas
def create_tf_matrix(freq_matrix):
    tf_matrix = {}
    for sent, f_table in freq_matrix.items():
        tf_table = {}
        count_words_in_sentence = len(f_table)
        for word, count in f_table.items():
            tf_table[word] = count / count_words_in_sentence if count_words_in_sentence > 0 else 0
        tf_matrix[sent] = tf_table
    return tf_matrix

tf_matrix = create_tf_matrix(freq_matrix)

print("--- 2. PERHITUNGAN TF (TERM FREQUENCY) LATIHAN ---")
for sent_key, tf in tf_matrix.items():
    print(f"Kalimat (slice): {sent_key}...")
    print(tf)
    print()
```

```python
import pandas as pd
import numpy as np
import re
```

```python
teks_tugas = (
    "Dinas Kesehatan Kabupaten/Kota dan Puskesmas juga dapat membuat pos pelayanan vaksinasi COVID-19. "
    "Dianjurkan agar setiap sasaran mencari informasi terlebih dahulu terkait jadwal layanan masing-masing fasilitas pelayanan kesehatan atau pos pelayanan vaksinasi"
)

# Membagi teks menjadi 2 kalimat berdasarkan titik
kalimat_list = [k.strip() for k in teks_tugas.split('.') if k.strip()]

print("Daftar Kalimat:")
for i, k in enumerate(kalimat_list, 1):
    print(f"Kalimat {i}: {k}")
print("\n" + "="*50 + "\n")
```

```python
# 1. Membuat Matriks Frekuensi Kata untuk Kalimat Tugas
def create_frequency_matrix_id(sentences):
    frequency_matrix = {}
    for i, sent in enumerate(sentences):
        freq_table = {}
        # Tokenisasi sederhana menggunakan regex untuk mengambil kata saja
        words = re.findall(r'\b\w+\b', sent.lower())
        for word in words:
            if word in freq_table:
                freq_table[word] += 1
            else:
                freq_table[word] = 1
        frequency_matrix[f"Kalimat {i+1}"] = freq_table
    return frequency_matrix

freq_matrix_tugas = create_frequency_matrix_id(kalimat_list)

print("--- 1. MATRIKS FREKUENSI KATA (TUGAS) ---")
for k, v in freq_matrix_tugas.items():
    print(f"{k}: {v}")
print("\n" + "="*50 + "\n")
```

```python
# 2. Perhitungan TF-IDF menggunakan scikit-learn
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(kalimat_list)

# Konversi hasil ke DataFrame agar rapi seperti tabel
df_tfidf = pd.DataFrame(
    X.toarray(),
    columns=vectorizer.get_feature_names_out(),
    index=[f"Kalimat {i+1}" for i in range(len(kalimat_list))]
)

print("--- 2. PERHITUNGAN TF-IDF (TUGAS) ---")
print(df_tfidf.T) # Ditranspose (.T) agar daftar kata berada di baris sebelah kiri
```

```python
import math
import string
import nltk
from nltk import sent_tokenize, word_tokenize
from nltk.stem import PorterStemmer
from nltk.corpus import stopwords
```

```python
# Unduh resource yang dibutuhkan
nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)

# a. Inisialisasi Teks
text_str = '''
Jumlah kebutuhan vaksin untuk program vaksinasi dalam penanganan
pandemi COVID-19 untuk mencapai herd immunity di Indonesia sangat
besar. Untuk itu, Pemerintah mengupayakan ketersediaan vaksin dari
berbagai sumber, salah satunya melalui kerja sama dengan negara lain.
Dalam mendukung kebijakan penyediaan vaksin COVID-19 tersebut, sebagai
Regulator Obat di Indonesia Badan POM melakukan pengawalan terhadap
pemenuhan Khasiat, Keamanan dan Mutu obat agar masyarakat dapat
mengakses Vaksin COVID-19 yang memenuhi standar dan persyaratan dan
dalam waktu yang tepat dengan menerbitkan Izin Penggunaan
Darurat/Emergency Use Authorization.
Sebelumnya, Badan POM telah mengeluarkan Izin Penggunaan
Darurat/Emergency Use Authorization (EUA) terhadap 7 produk vaksin
COVID-19, yaitu Vaksin CoronaVac (Sinovac), Vaksin COVID-19 Bio Farma,
Vaksin AstraZeneca, Vaksin Sinopharm, Vaksin Moderna, Vaksin Comirnaty
(Pfizer and BioNTech), dan Vaksin Sputnik-V. Selasa (07/09), Badan POM
kembali menerbitkan EUA bagi 2 (dua) produk vaksin COVID-19 yang baru,
yaitu Janssen COVID-19 Vaccine dan Vaksin Convidecia.
'''

sentences = sent_tokenize(text_str)
total_documents = len(sentences)
print("--- Output Tokenizing Kalimat ---")
print(sentences)
print("\n" + "="*50 + "\n")
```

```python
# b. Membuat matriks frekuensi kata-kata
def _create_frequency_matrix(sentences):
    frequency_matrix = {}
    stopWords = set(stopwords.words("indonesian"))
    ps = PorterStemmer()

    for sent in sentences:
        freq_table = {}
        words = word_tokenize(sent)
        for word in words:
            word = word.lower()
            if word in stopWords:
                continue

            if word in freq_table:
                freq_table[word] += 1
            else:
                freq_table[word] = 1

        frequency_matrix[sent[:15]] = freq_table
    return frequency_matrix

freq_matrix = _create_frequency_matrix(sentences)
```

```python
# c. Menghitung Matriks TF
def _create_tf_matrix(freq_matrix):
    tf_matrix = {}
    for sent, f_table in freq_matrix.items():
        tf_table = {}
        count_words_in_sentence = len(f_table)
        for word, count in f_table.items():
            tf_table[word] = count / count_words_in_sentence
        tf_matrix[sent] = tf_table
    return tf_matrix

tf_matrix = _create_tf_matrix(freq_matrix)
```

```python
# d. Tabel sederhana dokumen per kata
def _create_documents_per_words(freq_matrix):
    word_per_doc_table = {}
    for sent, f_table in freq_matrix.items():
        for word, count in f_table.items():
            if word in word_per_doc_table:
                word_per_doc_table[word] += 1
            else:
                word_per_doc_table[word] = 1
    return word_per_doc_table

count_doc_per_words = _create_documents_per_words(freq_matrix)
```

```python
# e. Menghitung Matriks IDF
def _create_idf_matrix(freq_matrix, count_doc_per_words, total_documents):
    idf_matrix = {}
    for sent, f_table in freq_matrix.items():
        idf_table = {}
        for word in f_table.keys():
            idf_table[word] = math.log10(total_documents / float(count_doc_per_words[word]))
        idf_matrix[sent] = idf_table
    return idf_matrix

idf_matrix = _create_idf_matrix(freq_matrix, count_doc_per_words, total_documents)
```

```python
# f. Menghitung Matriks TF-IDF
def _create_tf_idf_matrix(tf_matrix, idf_matrix):
    tf_idf_matrix = {}
    for (sent1, f_table1), (sent2, f_table2) in zip(tf_matrix.items(), idf_matrix.items()):
        tf_idf_table = {}
        for (word1, value1), (word2, value2) in zip(f_table1.items(), f_table2.items()):
            tf_idf_table[word1] = float(value1 * value2)
        tf_idf_matrix[sent1] = tf_idf_table
    return tf_idf_matrix

tf_idf_matrix = _create_tf_idf_matrix(tf_matrix, idf_matrix)
```

```python
# g. Penskoran kalimat
def _score_sentences(tf_idf_matrix) -> dict:
    sentenceValue = {}
    for sent, f_table in tf_idf_matrix.items():
        total_score_per_sentence = 0
        count_words_in_sentence = len(f_table)
        for word, score in f_table.items():
            total_score_per_sentence += score

        if count_words_in_sentence > 0:
            sentenceValue[sent] = total_score_per_sentence / count_words_in_sentence
        else:
            sentenceValue[sent] = 0
    return sentenceValue

sentence_scores = _score_sentences(tf_idf_matrix)
```

```python
# h. Menghitung rata-rata
def _find_average_score(sentenceValue) -> int:
    sumValues = 0
    for entry in sentenceValue:
        sumValues += sentenceValue[entry]
    average = (sumValues / len(sentenceValue))
    return average

threshold = _find_average_score(sentence_scores)
```

```python
# i. Hasil Ringkasan
def _generate_summary(sentences, sentenceValue, threshold):
    sentence_count = 0
    summary = ''
    for sentence in sentences:
        if sentence[:15] in sentenceValue and sentenceValue[sentence[:15]] >= (threshold):
            summary += " " + sentence
            sentence_count += 1
    return summary

summary = _generate_summary(sentences, sentence_scores, 1.3 * threshold)
print("--- Output Summary ---")
print(summary)
```
