```python
import math, re
import numpy as np
import pandas as pd

docs = ["makin kesini semakin susah buat dapetin drivernya",
        "aplikasi aneh, masa mau minta batalin pesanan food gak bisa",
        "padahal menu yg dipesan salah",
        "bagus pelayannya ramah Drivernya",
        "dapat harga murah malah di komplain sama driver nya 🤣",]

N = len(docs)
tok = [d.split() for d in docs]
vocab = sorted({w for t in tok for w in t})
tf = pd.DataFrame([[t.count(w) for w in vocab] for t in tok], columns=vocab, index=["d1", "d2", "d3", "d4", "d5"])

df_ = (tf > 0).sum()
idf = np.log10(N / df_)

print("TF (frekuensi mentah):"); print(tf.to_string())
print("\nDF :", df_.to_dict())
print("IDF:", idf.round(3).to_dict())
```

```python
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

cv = CountVectorizer()
X = cv.fit_transform(docs)
print(type(X), X.shape)
print(pd.DataFrame(X.toarray(), columns=cv.get_feature_names_out(), index=tf.index).to_string())
```

```python
tfidf_manual = tf * idf
print(tfidf_manual.round(3).to_string())
```

```python
tv = TfidfVectorizer()
Xt = tv.fit_transform(docs)
tfidf_sk = pd.DataFrame(Xt.toarray(), columns=tv.get_feature_names_out(), index=tf.index)
print(tfidf_sk.round(3).to_string())
print("\nIDF sklearn:", {w: round(float(v), 3) for w, v in zip(tv.get_feature_names_out(), tv.idf_)})
```

```python
def top_terms(row, k=3):
  return row.sort_values(ascending=False).head(k).round(3).to_dict()
for d in tfidf_sk.index:
  print(d, top_terms(tfidf_sk.loc[d]))
```

```python
import re, math, time
import numpy as np
import pandas as pd
berita = pd.read_csv("/content/drive/MyDrive/berita_daring_20.csv")          # 20 judul (P1, contoh sintetis)
docs = dict(zip(berita["id"], berita["judul"].str.lower()))
def token(t):
    return re.findall(r"[a-z]+", t.lower())
index = {}
for d, t in docs.items():
    for w in set(token(t)):
        index.setdefault(w, set()).add(d)
print("jumlah term:", len(index))
for w in ["daring", "kuota", "sinyal", "tatap"]:
    print(f"{w:7} df={len(index[w]):2}  {sorted(index[w])}")
```

```python
def jaccard(a, b):
    A, B = set(token(a)), set(token(b))
    return len(A & B) / len(A | B)
def levenshtein(s, t):
    prev = list(range(len(t) + 1))
    for i, cs in enumerate(s, 1):
        cur = [i]
        for j, ct in enumerate(t, 1):
            cur.append(min(prev[j] + 1, cur[j - 1] + 1, prev[j - 1] + (cs != ct)))
        prev = cur
    return prev[-1]
print("jaccard :", round(jaccard(docs["B02"], docs["B03"]), 3))
for a, b in [("daring", "darling"), ("kuliah", "kulia"), ("sinyal", "signal"), ("zoom", "room")]:
    print(f"lev({a},{b}) = {levenshtein(a, b)}")
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
ids = list(docs); teks = list(docs.values())
tv = TfidfVectorizer()
X = tv.fit_transform(teks)
def cari_tfidf(q, k=5):
    s = cosine_similarity(tv.transform([q]), X).ravel()
    urut = s.argsort()[::-1][:k]
    return [(ids[i], round(float(s[i]), 3)) for i in urut if s[i] > 0]
print(cari_tfidf("sinyal internet desa"))
```

```python
from rank_bm25 import BM25Okapi
bm25 = BM25Okapi([token(t) for t in teks], k1=1.5, b=0.75)
def cari_bm25(q, k=5):
    s = bm25.get_scores(token(q))
    urut = s.argsort()[::-1][:k]
    return [(ids[i], round(float(s[i]), 3)) for i in urut if s[i] > 0]
print(cari_bm25("sinyal internet desa"))
```
